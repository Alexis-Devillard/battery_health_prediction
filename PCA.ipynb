PCA on the battery health prediction

```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

Creating the data frame and visualising the content

```python
data = pd.read_csv("dataset/ev battery_failure prediction Dataset.csv")
data.head(10)
```

```python
data.describe()
```

Separate the qualitative and quantitatives data

```python
# Separate battery_failure which is a boolean and not a quantitative value
data['battery_failure'] = data['battery_failure'].astype('category')

# Separate data
df_quanti = data.select_dtypes(include='number')
df_quali = data.select_dtypes(exclude='number')

# Fill the NaN in the data (necessary for the next steps)
df_quanti = df_quanti.fillna(df_quanti.median())

df_quanti.head(2)
```

Standardisation : Centering and scaling the quantitative data

```python
stand_data = StandardScaler().fit_transform(df_quanti)
print(stand_data)
```

```python
# Check that the means are 0 and the standard deviations are 1
print(np.mean(stand_data, axis=0))
print(np.std(stand_data, axis=0))
```

Apply the PCA

```python
# Instantiate the PCA (all components are kept by default)
acp = PCA()
# Run the PCA and get the factorial coordinates of the individuals in the matrix coord
coord = acp.fit_transform(stand_data)
print(coord)
# Number of computed components
n = acp.n_components_
print(n)
```

Eigen value and inertia

```python
# Get the eigenvalues / variances / inertia
# sklearn estimates the variance with n-1 degrees of freedom
variance = acp.explained_variance_
variance
```

```python
# An adjustment is needed to get the variance with n degrees of freedom (the usual convention in PCA)
eigval = variance * (len(stand_data) - 1) / len(stand_data)
print(eigval)
```

```python
# Check that the sum of the eigenvalues equals the number of variables
print("Number of variables: ", n)
print(np.sum(eigval, axis=0))
```

```python
# Share of each eigenvalue,
# i.e. percentage of inertia explained by each axis
np.set_printoptions(suppress=True)
print(acp.explained_variance_ratio_ * 100)
```
